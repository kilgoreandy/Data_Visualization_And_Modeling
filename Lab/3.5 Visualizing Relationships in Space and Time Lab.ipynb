# What is our data? 

The dataset comes from Our World in Data (OWID), a COVID-19 dataset. It contains country-level daily records of COVID-19 cases, deaths, vaccinations, demographics, and economic indicators.

Spatial Features: country, code (ISO country code), continent.

Temporal Feature: date (daily records from 2020 onward).

COVID Metrics: total_cases, new_cases, total_deaths, new_deaths.

Context Features: population, gdp_per_capita, life_expectancy, etc.

This dataset allows us to track how the pandemic evolved over time (waves of infections) and across space (differences between countries and continents).
https://ourworldindata.org/coronavirus#explore-our-data-on-covid-19


# Visualization 1: Daily New Cases in the U.S.

```python

import pandas as pd
import plotly.express as px

# Load dataset
df = pd.read_csv("COVID Data.csv")
df.head()

df_country = df[df["country"] == "United States"]

fig = px.line(df_country, x="date", y="new_cases", 
              title="Daily New COVID-19 Cases in the United States")
fig.show()

```

Visualization 1 shows the daily new COVID-19 cases in the United States from 2020 through 2025 in millions. The graph highlights multiple waves of infection, with the largest surge occurring in early 2022. The jagged pattern reflects noise from daily fluctuations in reported cases, while the recurring peaks demonstrate seasonality, as waves of the virus rose and fell over time.

# Visualization 2: Total Cases per million over time


```python
fig = px.choropleth(df, 
                   locations="code",
                   color="total_cases_per_million",
                   hover_name="country",
                   animation_frame="date",
                   title="COVID-19 Total Cases per Million Over Time",
                   projection="natural earth",
                   color_continuous_scale="Reds")
fig.show()


```

This choropleth map shows how COVID-19 spread across countries over time. In early 2020, only a few regions are shaded which shows that the outbreak was local. As the time goes on, more countries darken in color, reflecting increasing cases per million. This visualization captures the spatial spread of the pandemic, highlighting how the virus transitioned from a few countries to a global crisis in the matter of a few years. 

# Visualization 3: Daily Deaths in Top Countries 

```python
countries = ["United States", "India", "Brazil", "United Kingdom", "South Africa"]
df_subset = df[df["country"].isin(countries)]

fig = px.line(df_subset, x="date", y="new_deaths", color="country",
              title="Daily New COVID-19 Deaths in Selected Countries")
fig.show()
```

This line chart compares daily new COVID-19 deaths across five top countries. Each country experienced peaks at different times, showing how the pandemic’s severity moved globally. While all countries show waves of deaths, the timing and scale differ, highlighting differences in impact across the globe.
